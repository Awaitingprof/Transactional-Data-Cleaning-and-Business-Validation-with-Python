```python
segment_shipmode = {
    "First Class": {
        "First Class": "Consumer",
        "Unnamed: 3": "Corporate",
        "Unnamed: 4": "Home Office"
    },
    "Same Day": {
        "Same Day": "Consumer",
        "Unnamed: 6": "Corporate",
        "Unnamed: 7": "Home Office"
    },
    "Second Class": {
        "Second Class": "Consumer",
        "Unnamed: 9": "Corporate",
        "Unnamed: 10": "Home Office"
    },
    "Standard Class": {
        "Standard Class": "Consumer",
        "Unnamed: 12": "Corporate",
        "Unnamed: 13": "Home Office"
    }
}
```

```python
segment_shipmode
```

```python
records = []

for ship_mode, columns in segment_shipmode.items():
    for column, segment in columns.items():

        temp = df_data[["Unnamed: 0", "Ship Mode", column]].copy()

        temp.columns = ["Order ID", "Order Date", "Sales"]

        temp["Ship Mode"] = ship_mode
        temp["Segment"] = segment

        records.append(temp)

long_df = pd.concat(records, ignore_index=True)

long_df = long_df[
    ["Order ID", "Order Date", "Ship Mode", "Segment", "Sales"]
]
```

```python
Q1 = long_df["Sales"].quantile(0.25)
Q3 = long_df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = long_df[long_df["Sales"] > upper_bound]

outliers.shape
```
